# 06 — Results Consolidation

Loads the tables produced by `src/experiments/exp1-exp6` (run them first via `python -m src.experiments.runner`) and builds a consolidated summary. It does not generate synthetic data — if an experiment has not run yet, it shows up as missing.

```python
import sys, os
sys.path.insert(0, '..')
import pandas as pd

TABLES_DIR = '../results/tables'
EXPECTED = {
    'exp1_baseline_comparison.csv': 'Baseline (7 systems)',
    'exp2_robustness.csv': 'Robustness by attack type',
    'exp3_scalability.csv': 'Scalability',
    'exp4_adaptive.csv': 'Adaptive adversary',
    'exp6_fairness.csv': 'Fairness across clients',
}

available = {}
for fname, label in EXPECTED.items():
    path = os.path.join(TABLES_DIR, fname)
    if os.path.exists(path):
        available[label] = pd.read_csv(path)
        print(f'✓ {label} ({fname})')
    else:
        print(f'✗ {label} ({fname}) — not found; run '
              f'`python -m src.experiments.runner` first')
```

```python
for label, df in available.items():
    print(f'\n=== {label} ===')
    display(df)
```
