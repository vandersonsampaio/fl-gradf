# 06 — Consolidação dos Resultados

Carrega as tabelas geradas por `src/experiments/exp1-exp6` (rode-os antes via `python -m src.experiments.runner`) e monta um resumo consolidado. Não gera dados sintéticos — se um experimento ainda não rodou, ele aparece como ausente.

```python
import sys, os
sys.path.insert(0, '..')
import pandas as pd

TABLES_DIR = '../results/tables'
EXPECTED = {
    'exp1_baseline_comparison.csv': 'Baseline (7 sistemas)',
    'exp2_robustness.csv': 'Robustez por tipo de ataque',
    'exp3_scalability.csv': 'Escalabilidade',
    'exp4_adaptive.csv': 'Adversário adaptativo',
    'exp6_fairness.csv': 'Fairness entre clientes',
}

available = {}
for fname, label in EXPECTED.items():
    path = os.path.join(TABLES_DIR, fname)
    if os.path.exists(path):
        available[label] = pd.read_csv(path)
        print(f'✓ {label} ({fname})')
    else:
        print(f'✗ {label} ({fname}) — não encontrado; rode '
              f'`python -m src.experiments.runner` primeiro')
```

```python
for label, df in available.items():
    print(f'\n=== {label} ===')
    display(df)
```
